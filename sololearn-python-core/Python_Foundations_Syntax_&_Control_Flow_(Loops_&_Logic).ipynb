```markdown
## Introduction to loops: for and while / Introduzione ai cicli: for e while

### Choosing the right Loop (The Golden Rule) / Scegliere il ciclo giusto (La Regola d'Oro)

**English:** Generally, use **for** loops when you already know the number of iterations (e.g., "do this 10 times"). Use **while** loops when there is a condition that must be met, and you don't know exactly when it will happen (e.g., "do this until the seats are empty").

**Italiano:** In generale, utilizza i cicli **for** quando conosci già il numero di iterazioni (es. "fai questo 10 volte"), e i cicli **while** quando c'è una condizione che deve essere soddisfatta e non sai esattamente quando accadrà (es. "fai questo finché i posti non sono esauriti").
```

````python
#Define the number of iterations
# Transliterated in Italian: "Per ogni numero (chiamato 'i' - abbreviazione di 'iteration') all'interno dell'intervallo da 0 a 2: esegui le istruzioni seguenti."
# Transliterated in English: "For each number (called 'i' - short for 'iteration') within the range from 0 to 2: execute the following instructions."
for i in range(3):
  #Statement that gets repeated
  print("Hello")
````

````python
# In questo ciclo, 'i' assume un valore diverso a ogni iterazione.
# 'range(10)' prepara i numeri da 0 a 9.

for i in range(10):
  # Qui NON usiamo le virgolette.
  # Python stampa il VALORE contenuto nella variabile 'i'.
  # Iterazione 1: i=0 -> stampa 0
  # Iterazione 2: i=1 -> stampa 1
  # ...e così via.
  print(i)
````

````python
# Loop using i as internal variable
for i in range(3):
  print(i)

print("--")   #This is just a separator

#Loop using "something" as internal variable
for something in range(3):
  print(something)
````

Per iterare la parola letterale "something" (scorrere ogni singolo carattere della parola), puoi usare il seguente codice:

````python
for char in "something":
  print(char)
````

Se invece intendi stampare la parola "something" (come stringa) più volte, puoi fare così:

````python
for _ in range(3):
  print("something")
````

### Prossimi passi: Logica e Applicazioni / Next steps: Logic and Applications

**Italiano:** Ora che abbiamo visto i cicli, approfondiamo la logica che li muove.
**English:** Now that we've seen loops, let's dive deeper into the logic that drives them.

````python
print(5 == 5) # is 5 equal to 5? / 5 è uguale a 5?
print(5 == 7) # is 5 equal to 7? / 5 è uguale a 7?
print(5 != 7) # is 5 different from 7? / 5 è diverso da 7?
print(5 != 5) # is 5 different from 5? / 5 è diverso da 5?
print(5 <= 5) # is 5 less than or equal to 5? / 5 è minore o uguale a 5?
print(5 >= 5) # is 5 greater than or equal to 5? / 5 è maggiore o uguale a 5?
````

````python
# Esempio di ciclo while
count = 0
while count < 5:
  print(f"Il conteggio è: {count}")
  count += 1 # Questo è fondamentale per evitare un loop infinito!

print("Ciclo while terminato.")
````

### Comparison of the three While approaches / Confronto dei tre approcci While

1. **Standard While (`while cond:`)**
   - **English**: Checks the condition *before* acting. It stops as soon as the condition is false.
   - **Italiano**: Controlla la condizione *prima* di agire. Si ferma appena la condizione è falsa.

2. **While with Feedback (`while cond:` + final `print`)**
   - **English**: Same logic as the standard one, but informs the user when the process is finished.
   - **Italiano**: Stessa logica dello standard, ma informa l'utente quando il processo è terminato.

3. **Controlled Infinite While (`while True:` + `break`)**
   - **English**: Always enters the loop and checks for the exit condition *inside* using `break`.
   - **Italiano**: Entra sempre nel ciclo e controlla la condizione di uscita *all'interno* usando `break`.

### Example: Ticket Sales / Esempio: Vendita Biglietti

**Loop Instruction:**
- **English:** "While seats are greater than zero, execute the following instructions."
- **Italiano:** "Finché i posti sono maggiori di zero, esegui le istruzioni seguenti."

```markdown
### 2. While with Feedback / While con Feedback

**English:** Same logic as the standard one, but we add a notification (print) at the end of the process.
**Italiano:** Stessa logica dello standard, ma aggiungiamo una notifica (print) alla fine del processo.
```

````python
seats = 500  # initial number of seats

while seats > 0:  # seat available?
  # print("Sell ticket") # ticket sold (commentato per non intasare l'output)
  seats = seats - 1  # number of seats updated

print(f"Tutti i biglietti sono stati venduti! Posti rimanenti: {seats}")
````

### Differences between While loops / Differenze tra i cicli While

1. **Standard While (`while cond:`)**:
   - **English**: Checks the condition *before* acting. It stops as soon as the condition is false.
   - **Italiano**: Controlla la condizione *prima* di agire. Si ferma appena la condizione è falsa.

2. **While with Feedback (`while cond:` + final `print`)**:
   - **English**: Same logic as the standard one, but informs the user when the process is finished.
   - **Italiano**: Stessa logica dello standard, ma informa l'utente quando il processo è terminato.

3. **Controlled Infinite While (`while True:` + `break`)**:
   - **English**: Always enters the loop and checks for the exit condition *inside* using `break`.
   - **Italiano**: Entra sempre nel ciclo e controlla la condizione di uscita *all'interno* usando `break`.

#### Type 1: Standard While / Tipo 1: While Standard

**English:** This matches the first description: it checks the condition at the top and stops when it is no longer true.
**Italiano:** Questo corrisponde alla prima descrizione: controlla la condizione in alto e si ferma quando non è più vera.

````python
seats = 500 # initial number of seats
while seats > 0: # seat available?
  print("Sell ticket") # ticket sold
  seats = seats - 1 # number of seats updated
````

### Example: Infinite Loops and 'break' / Esempio: Loop infiniti e 'break'

**Loop Instruction:**
- **English:** "While True, execute the following instructions, but stop if a specific condition is met."
- **Italiano:** "Finché è Vero (sempre), esegui le istruzioni, ma fermati se si verifica una condizione specifica."

```markdown
### 3. Controlled Infinite While / While Infinito Controllato

**English:** It uses `while True` to start an "infinite" loop and a `break` command inside to stop it when a condition is met.
**Italiano:** Usa `while True` per iniziare un ciclo "infinito" e un comando `break` all'interno per fermarlo quando una condizione è soddisfatta.
```

````python
seats = 10

while True: # Questo tecnicamente è un loop infinito / Always true
    if seats > 0:
        print(f"Biglietto venduto! Posti rimasti: {seats}")
        seats -= 1
    else:
        print("Posti esauriti!")
        break # Questo comando FERMA il ciclo immediatamente / Stops the loop

print("Il ciclo è stato interrotto.")
````

### Understanding the Counter / Capire il Contatore

**English:** In this example, `seats` is our counter. Every time we sell a ticket, we subtract 1. The loop stops when the counter reaches 0.

**Italiano:** In questo esempio, `seats` è il nostro contatore. Ogni volta che vendiamo un biglietto, sottraiamo 1. Il ciclo si ferma quando il contatore arriva a 0.

````python
seats = 5  # We start with 5 / Iniziamo con 5

while seats > 0:
    print(f"Selling ticket... Counter is now at: {seats}")

    # The Counter: This line is what eventually STOPS the loop
    # Il Contatore: Questa riga è ciò che alla fine FERMA il ciclo
    seats = seats - 1

print("Loop stopped because counter reached 0.")
````

### The Infinite Loop Trap / La Trappola del Loop Infinito

**English:** If we forget to update the counter, the condition never becomes `False`.
**Italiano:** Se dimentichiamo di aggiornare il contatore, la condizione non diventa mai `False`.

````python
# WARNING: This logic would run forever / Questa logica girerebbe all'infinito
# counter = 0
# while counter < 4:
#     print(counter)
#     # MISSING: counter = counter + 1

# Corrected version / Versione corretta:
counter = 0
while counter < 4:
    print(f"Correct loop: {counter}")
    counter = counter + 1 # Update the counter / Aggiorna il contatore
````

```markdown
## Boolean Logic and Password Application / Logica Booleana e Applicazione Password

**English:** Before the practical application, let's recap the comparison operators. These expressions return `True` or `False`.
**Italiano:** Prima dell'applicazione pratica, riepiloghiamo gli operatori di confronto. Queste espressioni restituiscono `True` o `False`.
```

````python
# Titolo: Esempio di Controllo Password / Example: Password Check

password = "SecretWord"
guess = "" # Inizializziamo la variabile / Initialize the variable

# Il ciclo continua finché guess è DIVERSO (!=) da password
# The loop continues as long as guess is DIFFERENT (!=) from password
while guess != password:
    guess = input("Inserisci la password: ")

print("Accesso consentito / Access granted")
````
