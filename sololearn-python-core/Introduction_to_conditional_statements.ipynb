Selection in programming can be likened to a **crossroads**. It is the mechanism that brings intelligence to your scripts, where the program decides which path to take based on the data provided.

### **Terminological Note: Conditional Statements**

In the Python ecosystem, these structures are formally known as **Conditional Statements**.

*   **Statement:** A fundamental command that the computer is instructed to execute.
*   **Conditional:** Implies that the execution of the statement is contingent upon whether a specified condition evaluates to `True` or `False`.

## **Neuropsychological Assessment**

In clinical neuropsychology, patient performance is rigorously compared against normative datasets to identify potential cognitive deficits.

> **Diagnostic Logic:** Using a standardized threshold to determine if a patient's memory performance suggests neurological impairment or remains within the expected cognitive range.

```python
# Neuropsychological test score (e.g., Digit Span Task)
patient_score = 4

# Normative threshold for cognitive health
normative_cutoff = 5
if patient_score >= normative_cutoff:

    # Performance is statistically within the expected range
    status = "Intact cognitive profile"

else:
    # Performance suggests a potential neurological deficit
    status = "Impaired cognitive profile"

print(f"Neuropsychological Status: {status}")
```

Comparison operators are the mathematical tools used to evaluate relationships between variables.

`==` `!=` `>` `<` `>=` `<=`

### **The 'Not Equal' Operator (`!=`)**

This operator verifies if two values are distinct. In experimental design, this is frequently used to exclude specific groups or filter non-target stimuli.

*   `==` checks for equality.
*   `!=` checks for inequality.

> **Laboratory Example:** Verifying if a participant's assigned group is distinct from the 'Control' group to trigger treatment-specific protocols.

```python
# Example: Group assignment check
participant_group = "Experimental"

# Check if the participant is NOT in the Control group
if participant_group != "Control":
    # This code runs if the group is Experimental, Pilot, etc.
    print("Participant receives the treatment.")
else:
    # This code runs only if the group is exactly "Control"
    print("Participant receives a placebo.")
```

### **The 'Equal To' Operator (`==`)**

The `==` operator is used to determine if two values are identical.

> **Technical Warning:** Do not confuse `==` (comparison) with `=` (assignment). The former evaluates a condition, while the latter stores a value in a variable.

```python
# Example: Protocol verification
# Imagine we are checking the modality of a brain scan
scan_modality = "fMRI"

# Check if the modality is EQUAL to fMRI
if scan_modality == "fMRI":
    # This code runs only if the string is exactly "fMRI"
    print("Applying Blood-Oxygen-Level Dependent (BOLD) analysis.")
else:
    # This runs for any other modality (e.g., PET, EEG, Structural MRI)
    print("Applying standard anatomical analysis.")
```

## **Handling Multiple Conditions: `elif`**

You can use the `elif` statement (short for "else if") to check additional conditions if the first `if` condition is not met.

> **Logical Flow:** This creates a chain of checks that stops as soon as one condition is True, optimizing execution speed in large datasets.

```python
# Example of multiple discount tiers
age = 75

if age < 18:
    # Checked first
    print("Junior discount applied")
elif age >= 75:
    # Checked only if age was NOT < 18
    print("Senior discount applied")
else:
    # Executed only if both above are False
    print("No discount applied")

print("Proceed to payment")
```

## **Nested Conditional Statements**

Sometimes, you need to check a condition, and *then* check another condition based on the result of the first. This is called **nesting**.

> **Case Study:** In a laboratory setting, this could mean first checking a participant's age, and then, only for minors, verifying parental consent before proceeding with the protocol.

```python
participant_age = int(input("Age assignment: "))
has_parental_consent = input("Confirm parental consent (yes/no): ")

# Protocol conditions assessment and assignment:
if participant_age < 18:
    # If the participant is under 18 years old
    if has_parental_consent.lower() == "yes": # Converts input to lowercase for robust comparison
        # If minor AND has parental consent
        print("Special protocol for minors with parental consent.")
    else:
        # If minor AND does NOT have parental consent
        print("Parental consent required. Participant cannot proceed.")
else:
    # If the participant is 18 years or older
    print("Standard protocol for adults.")
```
